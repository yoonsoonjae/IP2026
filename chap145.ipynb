# Morphological Transformations

## Erosion

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv2.imread('charj.png',0)
plt.subplot(121),plt.imshow(img)
plt.title('Orignal'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
img = cv2.imread('charj.png',0)
kernel = np.ones((5,5),np.uint8)
erosion1 = cv2.erode(img,kernel,iterations = 1)
erosion2 = cv2.erode(img,kernel,iterations = 2)
erosion3 = cv2.erode(img,kernel,iterations = 3)

plt.subplot(131),plt.imshow(erosion1)
plt.title('erosion1'), plt.xticks([]), plt.yticks([])
plt.subplot(132),plt.imshow(erosion2)
plt.title('erosion2'), plt.xticks([]), plt.yticks([])
plt.subplot(133),plt.imshow(erosion3)
plt.title('erosion3'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
img = cv2.imread('charj.png',0)
kernel = np.ones((5,5),np.uint8)
dilation1 = cv2.dilate(img,kernel,iterations = 1)
dilation2 = cv2.dilate(img,kernel,iterations = 2)
dilation3 = cv2.dilate(img,kernel,iterations = 3)

plt.subplot(131),plt.imshow(dilation1)
plt.title('dilation1'), plt.xticks([]), plt.yticks([])
plt.subplot(132),plt.imshow(dilation2)
plt.title('dilation2'), plt.xticks([]), plt.yticks([])
plt.subplot(133),plt.imshow(dilation3)
plt.title('dilation3'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
img = cv2.imread('charj.png',0)
kernel = np.ones((5,5),np.uint8)
opening = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)
plt.subplot(121),plt.imshow(opening)
plt.title('opening'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(closing)
plt.title('closing'), plt.xticks([]), plt.yticks([])
plt.show()
```
